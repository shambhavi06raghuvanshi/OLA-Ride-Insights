```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

df = pd.read_excel("../Dataset/ola_dataset.xlsx")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
print("Booking Status:")
print(df["Booking_Status"].value_counts())

print("\nVehicle Types:")
print(df["Vehicle_Type"].value_counts())

print("\nPayment Methods:")
print(df["Payment_Method"].value_counts())

print("\nIncomplete Ride Reasons:")
print(df["Incomplete_Rides_Reason"].value_counts())

print("\nCustomer Cancellation Reasons:")
print(df["Canceled_Rides_by_Customer"].value_counts())

print("\nDriver Cancellation Reasons:")
print(df["Canceled_Rides_by_Driver"].value_counts())
```

```python
print("Booking Status:")
print(df["Booking_Status"].value_counts())

print("\nVehicle Types:")
print(df["Vehicle_Type"].value_counts())

print("\nPayment Methods:")
print(df["Payment_Method"].value_counts())

print("\nIncomplete Ride Reasons:")
print(df["Incomplete_Rides_Reason"].value_counts())

print("\nCustomer Cancellation Reasons:")
print(df["Canceled_Rides_by_Customer"].value_counts())

print("\nDriver Cancellation Reasons:")
print(df["Canceled_Rides_by_Driver"].value_counts())
```

```python
print("Incomplete Rides:")
print(df["Incomplete_Rides"].value_counts(dropna=False))

print("\nDriver Ratings Summary:")
print(df["Driver_Ratings"].describe())

print("\nCustomer Ratings Summary:")
print(df["Customer_Rating"].describe())
```

```python

# Create a copy of the original dataset
df_clean = df.copy()

# Remove extra spaces from text columns
text_columns = df_clean.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    df_clean[col] = df_clean[col].str.strip()

# Confirm cleaning
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)
print("Duplicate rows:", df_clean.duplicated().sum())

print("\nCleaning completed successfully!")
```

```python

# Create a copy of the original dataset
df_clean = df.copy()

# Select only columns containing string values
text_columns = df_clean.select_dtypes(
    include=["object", "string"]
).columns

# Remove extra spaces from string values only
for col in text_columns:
    df_clean[col] = df_clean[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# Confirm cleaning
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)
print("Duplicate rows:", df_clean.duplicated().sum())

print("\nCleaning completed successfully!")
```

```python

# Save cleaned dataset as an Excel file
df_clean.to_excel("../Dataset/ola_cleaned.xlsx", index=False)

print("Cleaned dataset saved successfully!")
```

```python

# Count bookings by status
booking_counts = df_clean["Booking_Status"].value_counts()

print(booking_counts)

# Create a bar chart
plt.figure(figsize=(9, 5))
sns.barplot(x=booking_counts.index, y=booking_counts.values)

plt.title("Booking Status Distribution")
plt.xlabel("Booking Status")
plt.ylabel("Number of Bookings")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python

# Count bookings by date
daily_rides = df_clean.groupby("Date")["Booking_ID"].count()

# Create line chart
plt.figure(figsize=(12, 5))
plt.plot(daily_rides.index, daily_rides.values)

plt.title("Daily Ride Volume")
plt.xlabel("Date")
plt.ylabel("Number of Rides")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# Calculate total ride distance by vehicle
vehicle_distance = df_clean.groupby(
    "Vehicle_Type"
)["Ride_Distance"].sum().sort_values(ascending=False)

print(vehicle_distance)

# Create bar chart
plt.figure(figsize=(10, 5))
sns.barplot(
    x=vehicle_distance.index,
    y=vehicle_distance.values
)

plt.title("Total Ride Distance by Vehicle Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Total Ride Distance")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python

# Calculate average customer rating by vehicle
avg_ratings = df_clean.groupby(
    "Vehicle_Type"
)["Customer_Rating"].mean().sort_values(ascending=False)

print(avg_ratings)

# Create bar chart
plt.figure(figsize=(10, 5))
sns.barplot(x=avg_ratings.index, y=avg_ratings.values)

plt.title("Average Customer Ratings by Vehicle Type")
plt.xlabel("Vehicle Type")
plt.ylabel("Average Customer Rating")
plt.ylim(0, 5)
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python

# Calculate total revenue by payment method
payment_revenue = df_clean.groupby(
    "Payment_Method"
)["Booking_Value"].sum().sort_values(ascending=False)

print(payment_revenue)

# Create bar chart
plt.figure(figsize=(9, 5))
sns.barplot(x=payment_revenue.index, y=payment_revenue.values)

plt.title("Revenue by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Total Revenue")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python

# Count customer cancellation reasons
cancellation_counts = df_clean[
    "Canceled_Rides_by_Customer"
].value_counts()

print(cancellation_counts)

# Create bar chart
plt.figure(figsize=(11, 5))
sns.barplot(
    x=cancellation_counts.values,
    y=cancellation_counts.index
)

plt.title("Customer Ride Cancellation Reasons")
plt.xlabel("Number of Cancellations")
plt.ylabel("Cancellation Reason")
plt.tight_layout()
plt.show()
```

```python

# Count driver ratings
driver_rating_counts = df_clean["Driver_Ratings"].value_counts().sort_index()

print(driver_rating_counts)

# Create histogram
plt.figure(figsize=(9, 5))
sns.histplot(
    df_clean["Driver_Ratings"].dropna(),
    bins=10,
    kde=True
)

plt.title("Driver Ratings Distribution")
plt.xlabel("Driver Rating")
plt.ylabel("Number of Rides")
plt.tight_layout()
plt.show()
```

```python

# Compare customer and driver ratings
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df_clean,
    x="Customer_Rating",
    y="Driver_Ratings",
    alpha=0.4
)

plt.title("Customer Ratings vs Driver Ratings")
plt.xlabel("Customer Rating")
plt.ylabel("Driver Rating")
plt.tight_layout()
plt.show()
```

```python

# Filter successful bookings only
successful_rides = df_clean[
    df_clean["Booking_Status"] == "Success"
]

# Calculate revenue by payment method
payment_revenue = successful_rides.groupby(
    "Payment_Method"
)["Booking_Value"].sum().sort_values(ascending=False)

print("Revenue by Payment Method:")
print(payment_revenue)

# Plot chart
plt.figure(figsize=(9, 5))
sns.barplot(x=payment_revenue.index, y=payment_revenue.values)

plt.title("Revenue by Payment Method - Successful Rides")
plt.xlabel("Payment Method")
plt.ylabel("Total Revenue")
plt.tight_layout()
plt.show()
```

```python

# Export cleaned dataset to CSV
df_clean.to_csv("../Dataset/ola_cleaned.csv", index=False)

print("CSV file created successfully!")
```
